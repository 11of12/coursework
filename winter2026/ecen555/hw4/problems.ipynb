Problem 3.17

A: Heterojunction N+p

```python
from scipy.constants import Boltzmann, elementary_charge
import numpy as np

from scipy.optimize import curve_fit

def iv_curve(V, I_0, ideality_factor):
    T=300
    eta = ideality_factor
    k = Boltzmann
    T = T
    q = elementary_charge

    return I_0 * (np.exp(q*V/(eta*k*T)) - 1)
```

```python
import numpy as np
import matplotlib.pyplot as plt

voltage = np.array([0.206, 0.244, 0.29, 0.322, 0.362, 0.412, 0.453, 0.485, 0.537, 0.576, 0.612, 0.662, 0.708])
current = np.array([1.03e-9, 2.07e-9, 5.20e-9, 10.3e-9, 20.7e-9, 52.8e-9, 105e-9, 192e-9, 515e-9, 1.02e-6, 2.03e-6, 4.89e-6, 10.2e-6])

popt, pcov = curve_fit(iv_curve, voltage, current, p0=[50e-12, 2.24])
print(f"I0 = {popt[0]}")
print(f"eta = {popt[1]}")

plt.scatter(voltage, current)

voltage = np.linspace(np.min(voltage), np.max(voltage), 1000)
# plt.plot(voltage, iv_curve(voltage, I_0=50e-12, ideality_factor=2.24))
plt.plot(voltage, iv_curve(voltage, I_0=popt[0], ideality_factor=popt[1]))
plt.title("Dataset 1")
plt.xlabel("voltage")
plt.ylabel("current")
```
