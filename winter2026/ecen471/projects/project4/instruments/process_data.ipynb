```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data
file_path = "archive/TinySOL_metadata.csv"
with open(file_path) as file:
    csvFile = pd.read_csv(file)
    df = pd.DataFrame(csvFile)
```

```python
import librosa
import librosa.display
from rich.progress import Progress
import time

audio_paths = df["Path"]

num_mfcc = 20

X = np.zeros((len(audio_paths), num_mfcc, num_mfcc))
with Progress() as p:
    t = p.add_task("Processing...", total=X.shape[0])
    for i, path in enumerate(audio_paths):
        y, sr = librosa.load(path)
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20, n_fft=1800)
        X[i, :] = mfccs[:, mfccs.shape[1]//2 - num_mfcc//2: mfccs.shape[1]//2 + num_mfcc//2]
        # print(mfccs.shape)
        p.update(t, advance=1)

```

```python
instruments = set(df["Instrument (in full)"])
labels_map ={instrument: i for i, instrument in enumerate(instruments)}

y = np.array([labels_map[instrument] for instrument in df["Instrument (in full)"]])
```

```python
import pickle

data = {
    "X": X,
    "y": y,
    "labels_map": labels_map,
}

pickle.dump(data, open('processed_data.pkl', 'wb'))
```
