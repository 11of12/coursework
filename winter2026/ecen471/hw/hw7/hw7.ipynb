```python
import numpy as np

@np.vectorize
def h(x):
    if x <= 0:
        return 0
    else:
        return x
    
@np.vectorize
def h_prime(x):
    if x <= 0:
        return 0
    else:
        return 1

W1 = np.array([
    [0.2, 0.44],
    [0.32, -0.1],
    [0.1, 0.17],
])

b1 = np.array([0.47, 0.2, -0.32])

W2 = np.array([
    [0.87, 0.04, 0.52],
])

b2 = np.array([0.3])

x1 = np.array([-0.14, 0.15])
q1 = h(x1.T@W1.T + b1.T)
print(q1)
print(q1.T@W2.T + b2.T)
```

# Problem 3

```python
def compute_pooling_layer_output(input, stride=2, method=np.max):
    output = np.zeros((input.shape[0]//stride, input.shape[1]//stride))
    for i in range(output.shape[0]):
        for j in range(output.shape[1]):
            output[i, j] = method(input[stride*i:stride*(i+1), stride*j:stride*(j+1)])
    return output

pooling_layer_input = np.array([
    [5, 2, 1, 0, 2, 4],
    [2, 3, 0, 3, 4, 2],
    [5, 7, 7, 4, 3, 4],
    [0, 4, 9, 0, 2, 3],
    [1, 3, 5, 2, 6, 2],
    [2, 2, 4, 3, 8, 0]
])

print(f"Using Max Pooling: \n{compute_pooling_layer_output(pooling_layer_input, stride=2, method=np.max)}")
print(f"Using Average Pooling: \n{compute_pooling_layer_output(pooling_layer_input, stride=2, method=np.mean)}")
```

## Layer 1

```python
# The weight tensor is of dimensions 
# (filter rows × filter columns × input channels × output channels)
filter_rows = 5
filter_columns = 5
input_channels = 1
output_channels = 4

num_weight_parameters = filter_rows*filter_columns*input_channels*output_channels
# The number of parameters corresponds to the number of elements in the 
# weight tensor plus the number of elements in the offset vector (there is one element for each output channel)
num_offset_parameters = output_channels

num_parameters_first_layer = num_weight_parameters + num_offset_parameters
print(f"Number of parameters in first layer: {num_parameters_first_layer}")

# With a stride of 1, each of the p=784 pixel in the input image will have a corresponding hidden unit in the first layer for each of the 4 output channels
p = 28**2
num_hidden_units_first_layer = p * 4
print(f"Number of hidden units in first layer: {num_hidden_units_first_layer}")
```

## Layer  2

It is a similar process for the next two layers

```python
# The weight tensor is of dimensions 
# (filter rows × filter columns × input channels × output channels)
filter_rows = 5
filter_columns = 5
input_channels = 4
output_channels = 8

num_weight_parameters = filter_rows*filter_columns*input_channels*output_channels
# The number of parameters corresponds to the number of elements in the 
# weight tensor plus the number of elements in the offset vector (there is one element for each output channel)
num_offset_parameters = output_channels

num_parameters_second_layer = num_weight_parameters + num_offset_parameters
print(f"Number of parameters in second layer: {num_parameters_second_layer}")

p = 28**2

# The filter is now a tensor filter rows × filter columns × input channels
stride = 2
num_elements_in_filter = filter_rows*filter_columns*input_channels

num_hidden_units_in_single_channel_of_previous_layer = num_hidden_units_first_layer/input_channels

# We simply need to divide by the stride along each dimension the stride is applied (2 dimensions),
# Thus we divide by 2^2
num_hidden_units_in_single_channel_of_current_layer = num_hidden_units_in_single_channel_of_previous_layer / 2**2

num_hidden_units_second_layer =  num_hidden_units_in_single_channel_of_current_layer* output_channels
print(f"Number of hidden units in second layer: {num_hidden_units_second_layer}")
```
